# Challenge 10: Recursion

Complete a factorial function using recursion



````python
#!/bin/python3

import math
import os
import random
import re
import sys

#
# Complete the 'factorial' function below.
#
# The function is expected to return an INTEGER.
# The function accepts INTEGER n as parameter.
#

def factorial(n):
    result = 1
    if n <= 1:
        result = 1
    else:
        for i in range(n, 1, -1):
            result = i * result
    return result

if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w') #file pointer. It is a variable that points to an open file object

    #THIS WONT WORK UNLESS I CRETE AN ENVIROMENT! CREATE IT!

    n = int(input().strip())
    if n in range(2,13):

        result = factorial(n)
    else:
        print('Insert a number between 2 and 12, included')

    fptr.write(str(result) + '\n')

    fptr.close()
````

# Challenge 11: Binary transformation

Given a base-10 integer, n, convert it to binary (base-2). Then find and print the base-10 integer denoting the maximum number of consecutive 1's in n's binary representation. When working with different bases, it is common to show the base as a subscript.


````python
#!/bin/python3

import math
import os
import random
import re
import sys

def binary(n):
    binary_str = bin(n)[2:]
    longest_sequence = max(len(ones) for ones in binary_str.split("0"))
    print(int(longest_sequence))

if __name__ == '__main__':
    n = int(input().strip())
    if n >=1 and n <= 10 ** 6:
        binary(n)
    else:
        print('Not valid integer')
````

# Challenge 14: Abstract classes

Given a Book class and a Solution class, write a MyBook class that does the following:

Inherits from Book
Has a parameterized constructor taking these  parameters:
1. string title
2. string author
3. int price
Implements the Book class' abstract display() method so it prints these 3 lines:
Title: , a space, and then the current instance's title.
Author: , a space, and then the current instance's author .
Price: , a space, and then the current instance's price.
Note: Because these classes are being written in the same file, you must not use an access modifier (e.g.: public) when declaring MyBook or your code will not execute.

````python
from abc import ABCMeta, abstractmethod
class Book(object, metaclass=ABCMeta):
    def __init__(self,title,author):
        self.title=title
        self.author=author
    @abstractmethod
    def display(): pass

class MyBook(Book):
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price
    def display(self):
     print(f"Title: {self.title}")
     print(f"Author: {self.author}")
     print(f"Price: {self.price}")

title=input()
author=input()
price=int(input())
new_novel=MyBook(title,author,price)
new_novel.display()
````

A class constructor that takes an array of integers as a parameter and saves it to the __elements instance variable.
A computeDifference method that finds the maximum absolute difference between any 2 numbers in _-elements and stores it in the maximumDifference instance variable.

````python


class Difference:
  def __init__(self, a):
      self.__elements = a
  def computeDifference(self):
      Diff = []
      for i in range(len(self.__elements)):
          for j in range(len(self.__elements)):
              Diff.append(self.__elements[i] - self.__elements[j])
      self.maximumDifference = max(Diff)

# End of Difference class

_ = input() #quise no ponerlo con systdin y separarlo luego, pero me da un resultado muy raro
#parece que sin funciona en el entorno de HackerRank
a = [int(e) for e in input().split(' ')]

d = Difference(a)
d.computeDifference()

print(d.maximumDifference)
````
