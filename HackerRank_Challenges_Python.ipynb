Note: Because these classes are being written in the same file, you must not use an access modifier (e.g.: public) when declaring MyBook or your code will not execute.

````python
from abc import ABCMeta, abstractmethod
class Book(object, metaclass=ABCMeta):
    def __init__(self,title,author):
        self.title=title
        self.author=author
    @abstractmethod
    def display(): pass

class MyBook(Book):
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price
    def display(self):
     print(f"Title: {self.title}")
     print(f"Author: {self.author}")
     print(f"Price: {self.price}")

title=input()
author=input()
price=int(input())
new_novel=MyBook(title,author,price)
new_novel.display()
````

Some notes on time complexity using a primalty algorithm

First, lets write here the mathematical proof:

I a number n is not prime, it can be expressed as $n = a \times b$, where a and b are factors of n. If a number is not prime it can be factorized.

If both a and b are greater than $\sqrt{n}$, their product would exceed n. Therefore, a or b has to be less or equal than $\sqrt{n}$.

So, I start by looking to that divisor so I only have to range until $\sqrt{n}$ to find it. let's say it is a and n/a = b. By the mathematical proof b could be greater than $\sqrt{n}$, but then I assume b can be factorized as $b = c \times b$.

Time complexity reduces to $O(\sqrt{n})$



````python
def primis(data):
    if data == 1:
        return 'Not prime'
    if all(data % i != 0 for i in range(2, int(data ** 0.5) + 1)):
        return 'Prime'
    else:
      for i in range(2, int(data ** 0.5) + 1):
        if (data % i) == 0:
            return 'Not prime'
            break

datas = []
t = int(input())
for i in range(t):
    datas.append(int(input()))
for data in datas:
    ret = primis(data)
    print(ret)
````

Lets check carefully the time complexity of the code above:

- I have t test cases so the time complexity would be a product of t
- for each test case:
      - data == 1 is $O(1)$
      - data% i == 0 is $O(\sqrt{n})$
      - if all(data % i != 0 for i in range(2, int(data ** 0.5) + 1)) is $O(\sqrt{n})$
-------------------------------------------------------------------------------

TOTAL COMPLEXITY = $O(t\sqrt{n})$

# Instance, class and abstract methods!

- Instance methods are the most common. Belong to an instance of a class and can access the instance's attributes and methods (and also the class).
**self referes to the instance of the class and instance methods are called in an instance of the class**

- Class methods: declared with the @classmethod decorator

```
class MyClass:
    class_variable = 42

    @classmethod
    def class_method(cls):
        return cls.class_variable
```
- Static methods
the dont receive self or cls as the first parametor and therefore they cannot access instance or class attributes directly. You use it when you want to define behaviour independently of the isnstance or the class.

```
class MyClass:
    @staticmethod
    def static_method(x, y):
        return x + y
```

you can call these methods in both the instance or the class

Static methods are used for functions that are logically connected to the class but don't need access to instance or class attributes. Examples include:

  - Utility functions.
  - Helper functions that perform a task related to the class but don't modify the class state.

````python
class Math:
    constant = 3.14  # Class attribute

    def __init__(self, value):
        self.value = value  # Instance attribute

    # Instance method that depends on instance data
    def square(self):
        return self.value ** 2  # Operates on instance attribute

    # Class method that depends on class-level data
    @classmethod
    def multiply_with_constant(cls, number):
        return number * cls.constant  # Operates on class-level data

    # Static method that does not depend on instance or class data
    @staticmethod
    def add(x, y):
        return x + y  # Independent of instance/class data

# Create an instance of Math
math_instance = Math(4)

# Calling instance method (depends on instance data)
print(math_instance.square())  # Output: 16

# Calling class method (depends on class data)
print(Math.multiply_with_constant(5))  # Output: 15.7

# Calling static method (independent of instance or class data) (it can be also used creating an instance)
print(Math.add(3, 4))  # Output: 7
````

# Abstract methods

It is declared in an abstract class but not implemented. Subclasses are required to implement that abstract method. The abstract method serves as a contract that ensures that subclasses from abstract classes (their blueprint) inherit certain behaviour.

They are useful in situations where you want to define a common interface for multiple subclasses but expect each subclass to provide its own specific behavior.

Abstract classes can be defined with abstract and concrete methods

RegEx = Regular Expressions.

it is a way of characterizing a set of strings by a subset of common characteristics. The y add a level of abstraction to strings that makes it easier to search and manipulate text.

##Character Classes (not classes in thetraditional sense)
Regex character classes are set of characters or ranges of characters encolsed by square brackets [ ]

  - [abc] = matches a,b or c
  - [abc][pq] = matches a,b or c followed by either p or q
  - [^abc] = match any letter except a, b or c
  - [a-zA-Z] = match any lower or uppercase letter
  - [m-p2-8] = matches a letter between m and p and digits form 2 to 8 (but not p2!!!!!)
**Special characters**
The special characters are:
"." Matches any character except a newline. "^" Matches the start of the string. "$" Matches the end of the string or just before the newline at
 the end of the string.
"*" Matches 0 or more (greedy) repetitions of the preceding RE.
Greedy means that it will match as many repetitions as possible.
"+" Matches 1 or more (greedy) repetitions of the preceding RE. "?" Matches 0 or 1 (greedy) of the preceding RE. *?,+?,?? Non-greedy versions of the previous three special characters. {m,n} Matches from m to n repetitions of the preceding RE. {m,n}? Non-greedy version of the above. "\" Either escapes special characters or signals a special sequence. [] Indicates a set of characters.
 A "^" as the first character indicates a complementing set.
"|" A|B, creates an RE that will match either A or B. (...) Matches the RE inside the parentheses.

**Special sequences**:

- \A --> matches pattern at the start of the string = ^
- \Ƶ --> matches pattern at the end of the string = $
- \d --> matches to any digit = [0-9]
- \D --> matches to any nondigit = [^0-9]
- \s --> matches any whitespace = [\t\n\x0b\r\f]
- \S --> matches any NON whitespace = [^ \t\n\x0b\r\f]
- \w --> matches any alphanumeric character
- \W --> matches any NON alphanumeric character
- \b -->
Matches the empty string, but only at the beginning or end of a word. Matches a word boundary where a word character is [a-zA-Z0-9_].
For example, ‘\bJessa\b' matches ‘Jessa’, ‘Jessa.’, ‘(Jessa)’, ‘Jessa Emma Kelly’ but not ‘JessaKelly’ or ‘Jessa5’.
- \B -->
Opposite of a \b. Matches the empty string, but only when it is not at the beginning or end of a word

**Functions of the module**
- match Match a regular expression pattern to the beginning of a string.
- fullmatch Match a regular expression pattern to all of a string.
- search Search a string for the presence of a pattern.
- sub Substitute occurrences of a pattern found in a string. - subn Same as sub, but also return the number of substitutions made.
- split Split a string by the occurrences of a pattern.
- findall Find all occurrences of a pattern in a string.
- finditer Return an iterator yielding a Match object for each match.
- compile Compile a pattern into a Pattern object.
- purge Clear the regular expression cache.
- escape Backslash all non-alphanumerics in a string
